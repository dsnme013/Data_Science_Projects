```python
from langchain.chat_models import ChatOpenAI
from langchain.schema import AIMessage, HumanMessage, SystemMessage

# Create a ChatOpenAI instance with your desired settings
chat = ChatOpenAI(temperature=0.9, model='gpt-3.5-turbo', openai_api_key='')

# Define the maximum number of tokens for output
max_tokens = 400  # Adjust this value as needed

# Define a list to hold the conversation messages
messages = [
    SystemMessage(content="You are a helpful assistant"),
]

while True:
    # Get user input
    user_input = input("you: ")
    
    # Add user message to the list of messages
    messages.append(HumanMessage(content=user_input))
    
    # Generate AI response with increased token size
    ai_response = chat(messages=messages, max_tokens=max_tokens).content
    
    # Print AI response
    print("ai: ", ai_response)
    
    # Add AI message to the list of messages
    messages.append(AIMessage(content=ai_response))
```
